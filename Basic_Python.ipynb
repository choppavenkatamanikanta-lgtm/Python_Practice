```
def reverse(s):
  return s[::-1]
  print(reverse_string("python"))
```

```
def reverse_string(s):
  result = " "
  for ch in s:
    result = ch + result
    return result
```

```
def palindrome(s):
  return s == s [::-1]
  print(palindrome("madam"))
```

```
def fibonacci(n):
  a,b = 0, 1
  for _ in range(n):
    print(a, end=" ")
    a, b = b, a+b
    fibonacci(10)
```

```
def factorial(n):
  fact = 1
  for i in range(2, n + 1):
    fact *= i
    return fact
    print(factorial(5))
    def factorial(n):
      fact = 1
      for i in range(2, n + 1):
      fact *= i
      return fact
      print(factorial(5))
```
